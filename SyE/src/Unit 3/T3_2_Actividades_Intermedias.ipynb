## Actividad 7. Proyecto inicial

Crea un proyecto denominado **`agenda_personal`** que incluya:

```
agenda_personal/
├── README.md
├── .gitignore
└── main.py
```

El archivo `main.py` deberá mostrar un mensaje personalizado.

### 7.1 Crear la carpeta y los archivos

````python
from pathlib import Path

proyecto = Path("agenda_personal")
proyecto.mkdir(exist_ok=True)

for nombre in ["README.md", ".gitignore", "main.py"]:
    (proyecto / nombre).touch()

print("Contenido de la carpeta:")
for elemento in sorted(proyecto.iterdir()):
    print(" -", elemento.name)
````

### 7.2 Escribir `main.py` (personaliza el mensaje)

````python
%%writefile agenda_personal/main.py
nombre = "TU NOMBRE"
print(f"Hola, {nombre}. Bienvenido/a a tu agenda personal.")
````

### 7.3 Escribir `README.md` (complétalo)

````python
%%writefile agenda_personal/README.md
# Agenda personal

Describe aquí el objetivo del proyecto.

## Cómo ejecutarlo

    python main.py
````

### 7.4 Escribir `.gitignore`

````python
%%writefile agenda_personal/.gitignore
__pycache__/
.vscode/
.venv/
````

### 7.5 Ejecutar el proyecto

````python
!cd agenda_personal && python main.py
````

- Para poder organizar mejor los proyectos y que no se mezclen.
- - **.gitignore:** Especifica archivos que no deberán subirse a git/github.
- - **.README.md:** Es un archivo que resume el motivo del proyecto y suele tener la información de como ejecutar el proyecto.
- - **main.py:** Es el archivo principal del proyecto y el que ejecuta el resto del proyecto.

✍️ **Explica con tus palabras:**

| Comando | ¿Qué función cumple? |
|---|---|
| `git init` | Inicializa un proyecto con git. |
| `git add .` | Añade a la cola los archivos modificados para prepararlos para hacer un commit. |
| `git commit -m "Versión inicial"` | Crea una versión del proyecto para usarlo como un backup en caso de necesitarlo. |

Pregunta extra: ¿qué diferencia hay entre `git status` antes y después del `git add .`?

- Git status comprueba como están los archivos en ese momento.
- Antes del git add . salen en rojo porque no están listos para hacer commit.
- Después del git add . salen en verde porque indican que ya están listos para hacer el commit.
