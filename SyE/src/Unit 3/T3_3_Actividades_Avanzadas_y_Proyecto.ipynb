Los desarrollos de proyectos, por pequeños que sean, necesitan desarrollarse de manera profesional, para ello tienen un editor de texto estandarizado con un ecosistema grande y seguro, además, todo el código que programes necesita pasar los tests de calidad de código y de que los fallos esperados pasen como deben y los que deberían funcionar, que funcionen correctamente.

# 🧩 Proyecto integrador
## Creación de un entorno profesional de trabajo Python

**Situación planteada:** una pequeña empresa acaba de contratar a un desarrollador junior. El responsable técnico desea verificar que el nuevo empleado es capaz de preparar correctamente un entorno de desarrollo profesional antes de empezar a programar. **Tu tarea consiste en crear y documentar dicho entorno.**

### Requisitos mínimos

**Estructura del proyecto**

```
mi_primer_proyecto/
│
├── README.md
├── .gitignore
├── src/
│   └── main.py
└── tests/
```

**Configuración.** El proyecto deberá:

- Abrirse correctamente en VS Code.
- Tener seleccionado el intérprete adecuado.
- Incluir Git.
- Incluir documentación básica.
- Incluir carpetas `tests` y `src`.

**Prueba automatizada:** añadir una prueba sencilla utilizando `pytest`.

**Programa inicial:** `main.py` deberá mostrar el nombre del alumno, la fecha actual y un mensaje indicando que el entorno está operativo.

**Documentación obligatoria:** `README.md` con objetivo del proyecto, cómo ejecutarlo, herramientas utilizadas y comandos básicos necesarios.

### Entregables

- **Carpeta del proyecto** con todos los archivos requeridos.
- **Memoria técnica** (entre una y dos páginas): proceso seguido, problemas encontrados, soluciones aplicadas y aprendizajes obtenidos.

### P.1 Crear la estructura

````python
from pathlib import Path

raiz = Path("mi_primer_proyecto")
(raiz / "src").mkdir(parents=True, exist_ok=True)
(raiz / "tests").mkdir(exist_ok=True)

for ruta in ["README.md", ".gitignore", "pytest.ini", "src/main.py", "tests/test_main.py"]:
    (raiz / ruta).touch()

for elemento in sorted(raiz.rglob("*")):
    print(elemento.relative_to(raiz.parent))
````

### P.2 Programa inicial: `src/main.py`

Muestra nombre del alumno, fecha actual y mensaje de entorno operativo. **Escribe tu nombre.**

````python
%%writefile mi_primer_proyecto/src/main.py
"""Programa inicial del proyecto integrador."""
from datetime import date

NOMBRE_ALUMNO = "TU NOMBRE"


def construir_mensaje(nombre, fecha):
    """Devuelve el mensaje de bienvenida con nombre y fecha."""
    return (
        f"Alumno: {nombre}\n"
        f"Fecha: {fecha.strftime('%d/%m/%Y')}\n"
        "El entorno está operativo."
    )


def main():
    print(construir_mensaje(NOMBRE_ALUMNO, date.today()))


if __name__ == "__main__":
    main()
````

### P.3 Prueba automatizada: `tests/test_main.py`

````python
%%writefile mi_primer_proyecto/tests/test_main.py
from datetime import date

from main import construir_mensaje


def test_mensaje_contiene_nombre_fecha_y_estado():
    mensaje = construir_mensaje("Ana", date(2026, 1, 15))
    assert "Ana" in mensaje
    assert "15/01/2026" in mensaje
    assert "operativo" in mensaje
````

Para que pytest encuentre el código de `src/`, añadimos un pequeño archivo de configuración:

````python
%%writefile mi_primer_proyecto/pytest.ini
[pytest]
pythonpath = src
````
