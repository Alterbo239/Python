## Actividad 10. Dependency Hell

Investiga el concepto de Dependency Hell y explica:
- Qué significa.
- Qué problemas genera.
- Cómo ayudan los entornos virtuales a evitarlo.

Incluye ejemplos concretos.

- Es el infierno de dependencias, que consiste en que diferentes proyectos requieren de paquetes con versiones incompatibles, de manera que entran en conflicto y el entorno deja de ser reproducible.
- Esto genera que algunos proyectos no funcionen por depender de una versión incompatible o que en diferentes ordenadores, los proyectos funcionen distintos o que en alguno ni siquiera funcione. Esto provoca que se rompan algunos proyectos con errores inexplicables.
- En los entornos virtuales solo puedes tener una versión por paquete instalado, lo que evita completamente que hayan fallos de versiones. En caso de que en un ordenador funcione y en otro no, se deberá a que en un entorno la versión es la correcta y en el otro no, provocando que deba corregir la versión con la que trabajar.
