## Actividad 2. Entorno global vs entorno virtual

Explica las diferencias, ventajas e inconvenientes de cada opción:
- Opción A: Instalar todas las bibliotecas en el sistema global.
- Opción B: Utilizar entornos virtuales independientes.

### Respuesta

**A**
- Las bibliotecas instaladas podrían usarse en cualquier proyecto.
- No se duplicarían paquetes.
- Podrían haber fallos de versiones diferentes entre proyectos.
- Al estar globalizado, saber que paquetes utiliza cada proyecto se haría un mundo.

**B**
- Puedes gestionarlas individualmente y utilizar solo las necesarias para cada proyecto.
- Si dos o más proyectos requieren de un mismo paquete, tocaría instalarlo tantas veces como se necesite.
- Al tener cada proyecto sus propios paquetes, no hay conflictos.
- Es sencillo al crear el archivo requirements.txt sabes exacatamente qué paquetes utiliza cada proyecto.
