## Actividad 5. Exploración de objetos

Utiliza:

```python
texto = "Python"
dir(texto)
```

**Selecciona cinco métodos de la lista e investiga para qué sirven.**

````python
texto = "Python"
dir(texto)
````

Elige cinco métodos (sin guiones bajos), consulta su ayuda con `help(str.nombre_metodo)` y pruébalos. Ejemplo:

````python
help(str.upper)
print(texto.upper())
````

````python
print(texto.__contains__("Py"))
print(texto.__len__())
print(texto.__gt__("Zamora"))
print(texto.__eq__("python"))
print(texto.__hash__())

print(texto.upper())
print(texto.index("o"))
print(texto.isdigit())
print(texto.replace("thon", "Line"))
print(texto.rsplit("h"))
````

✍️ **Completa la tabla:**

| Método | ¿Para qué sirve? | Ejemplo y resultado |
|---|---|---|
| 1. __contains__ | Comprueba si un string contiene ciertos caracteres. | texto.__contains__("Py") -> True |
| 2. __len__ | Muestra la longitud del string. | texto.__len__() -> 6 |
| 3. __gt__ | Comprueba si el texto comienza con un caracter mayor según la tabla ASCII que el introducido. | texto.__gt__("Zamora") -> False |
| 4. __eq__ | Comprueba si una cadena es exactamente igual a otra. | texto.__eq__("python") -> False |
| 5. __hash__ | Devuelve un número que actúa como un ID del contenido de la variable. | texto.__hash__() -> -2109573180446508132 |

✍️ **Completa la tabla:**

| Método | ¿Para qué sirve? | Ejemplo y resultado |
|---|---|---|
| 1. .upper() | Es un método para poner en mayúsculas toda la cadena. | texto.upper() -> PYTHON |
| 2. .index() | Encuentra la posición de un caracter en la cadena. | texto.index("o") -> 4 |
| 3. .isdigit() | Muestra si la cadena es un número o no. | texto.isdigit() -> False |
| 4. .replace | Reemplaza la parte de la cadena del primer parámetro por el del segundo. | texto.replace("thon", "Line") -> PyLine |
| 5. .split() | Separa la cadena introducida por el caracter indicado en el parámetro y lo convierte en un array. (Esto afecta a tantas veces salga ese caracter introducido) | texto.rsplit("h") -> ['Pyt', 'on'] |
